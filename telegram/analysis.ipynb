```python
import json
import pandas as pd
import matplotlib.pyplot as plt

with open('output.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

data = list(data.values())
print(data)
df = pd.DataFrame(data)
print(df)
print(f'Всего сообщений: {len(df)}')
```

```python
top_users = df['user'].value_counts().head(10)

plt.figure(figsize=(10, 5))
top_users.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Топ-10 участников по количеству сообщений', fontsize=14)
plt.xlabel('Участник')
plt.ylabel('Количество сообщений')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

df['msg_len'] = df['text'].apply(len)
```

```python
avg_len = df.groupby('user')['msg_len'].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
avg_len.plot(kind='barh', color='lightcoral', edgecolor='black')
plt.title('Средняя длина сообщения по участникам', fontsize=14)
plt.xlabel('Средняя длина (количество символов)')
plt.ylabel('Участник')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
print('Общая статистика чата:')
print(f'Всего сообщений: {len(df)}')
print(f'Участников: {df['user'].nunique()}')
print(f'Средняя длина сообщения: {df['msg_len'].mean():.1f} символов')
print(f'Минимальная длина: {df['msg_len'].min()} символов')
print(f'Максимальная длина: {df['msg_len'].max()} символов')
print(f'Медианная длина: {df['msg_len'].median():.0f} символов')
```
